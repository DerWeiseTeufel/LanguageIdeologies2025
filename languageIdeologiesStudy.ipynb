```python
def find_word_context(text, word):
    """
    Find all occurrences of `word` in `text` and print each with
    10 preceding and 10 following characters.

    Args:
        text (str): The input text to search in.
        word (str): The word to find.
    """
    if not text or not word:
        print("Text or word is empty.")
        return

    word_len = len(word)
    text_len = len(text)
    occurrences = []

    # Use case-sensitive search; for case-insensitive: text.lower(), word.lower()
    start = 0
    while True:
        # Find next occurrence
        pos = text.find(word, start)
        if pos == -1:  # No more occurrences
            break

        # Get 10 chars before (or from start if less than 10)
        start_context = max(0, pos - 100)
        # Get 10 chars after (or until end if less than 10)
        end_context = min(text_len, pos + word_len + 100)

        context = text[start_context:end_context]
        occurrences.append(context)

        # Move past this occurrence
        start = pos + 1

    # Print all contexts
    for i, ctx in enumerate(occurrences, 1):
        print(f"{i}: {ctx}")




search_word = "putin"

find_word_context(all_text.lower(), search_word)
```

```python
import pandas as pd
import plotly.express as px
import spacy
from collections import Counter
import string

# Загрузка английской модели spaCy
nlp = spacy.load("en_core_web_sm")

# Пример датафрейма (замените на ваш источник данных)
# df = pd.read_csv("your_file.csv")  # или откуда вы берёте данные
df_sample = pd.DataFrame({
    "text": [
        "The quick brown fox jumps over the lazy dog.",
        "Python is a great programming language for data science.",
        "Data science combines statistics, programming, and domain expertise.",
        "Machine learning is a subset of artificial intelligence.",
        "Natural language processing helps computers understand human language."
    ]
})

# Функция для предобработки текста
def preprocess_text(text):
    # Приводим к нижнему регистру
    text = text.lower()
    # Удаляем пунктуацию
    text = text.translate(str.maketrans('', '', string.punctuation))
    # Обрабатываем текст с помощью spaCy
    doc = nlp(text)
    # Извлекаем леммы, фильтруем стоп‑слова и не-алфавитные токены
    lemmas = [
        token.lemma_ for token in doc
        if not token.is_stop and token.is_alpha
    ]
    return lemmas

# Применяем предобработку ко всей колонке text
all_lemmas = []
for text in df_sample["text"]:
    lemmas = preprocess_text(text)
    all_lemmas.extend(lemmas)

# Подсчитываем частоту слов
word_freq = Counter(all_lemmas)
most_common_100 = word_freq.most_common(100)

# Преобразуем в DataFrame для Plotly
freq_df = pd.DataFrame(most_common_100, columns=["word", "frequency"])

# Строим столбчатую диаграмму с помощью Plotly
fig = px.bar(
    freq_df,
    x="word",
    y="frequency",
    title="Top 100 Most Frequent Words",
    labels={"word": "Word", "frequency": "Frequency"},
    text="frequency",
    height=800
)

fig.update_traces(texttemplate="%{text}", textposition="outside")
fig.update_layout(
    xaxis=dict(tickangle=-45),  # поворот подписей по оси X для читаемости
    uniformtext_minsize=8,
    uniformtext_mode="hide"
)


# Отображаем график
fig.show()
```
